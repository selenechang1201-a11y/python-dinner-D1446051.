本程式會整理聚餐名稱、餐費與人數，計算服務費、折價後的總金額及每人分攤費用，並判斷是否超過每人 300 元的預算。

## 測試一：4 人，折價券 100 元


```python
# 原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 整理聚餐名稱：去除頭尾空白，再轉成小寫
clean_event_name = event_name.strip()
clean_event_name = clean_event_name.lower()

# 整理餐費文字，保留原始的 meal_amount
clean_meal_amount = meal_amount.replace("NT$", "")
clean_meal_amount = clean_meal_amount.replace(",", "")
clean_meal_amount = clean_meal_amount.replace(" ", "")

# 將餐費轉成數字，人數轉成整數
meal_cost = float(clean_meal_amount)
people_count = int(people)

# 先用原始餐費計算服務費，再扣除折價券
service_fee = meal_cost * service_rate
total_amount = meal_cost + service_fee - discount
cost_per_person = total_amount / people_count

# 判斷每人費用是否超過預算
over_budget = cost_per_person > budget_per_person

# 輸出費用摘要
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_cost:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{over_budget}")
```
